```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
dir(iris)
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['target']=iris.target
df
```

```python
df0=df[:50]
df1=df[50:100]
df2=df[100:]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.scatter(df0['sepal length (cm)'],df0['sepal width (cm)'],color='red',marker='x')
plt.scatter(df1['sepal length (cm)'],df1['sepal width (cm)'],color='black',marker='o')
plt.scatter(df2['sepal length (cm)'],df2['sepal width (cm)'],color='green',marker='*')
```

```python
X=df.drop('target', axis=1)
Y=df.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=1)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=10)
knn.fit(X_train,Y_train)
```

```python

knn.score(X_test,Y_test)
```

```python
from sklearn.metrics import confusion_matrix
y_pred=knn.predict(X_test)

cm=confusion_matrix(Y_test,y_pred)
cm
```

```python
import seaborn as sns

sns.heatmap(cm,annot=True)
```
