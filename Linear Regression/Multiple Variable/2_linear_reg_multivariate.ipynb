```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/codebasics/py/refs/heads/master/ML/2_linear_reg_multivariate/Exercise/hiring.csv")
df
```

Data Cleaning

```python
df['test_score(out of 10)']=df['test_score(out of 10)'].fillna(int(df['test_score(out of 10)'].mean()))
df
```

```python
!pip install word2number

```

```python
from word2number import w2n
def convert_experience(x):
  try:
    return w2n.word_to_num(x)
  except:
    return None
df['experience']=df['experience'].apply(convert_experience)
df
```

```python
df['experience']=df['experience'].fillna(0)
df
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[['experience','test_score(out of 10)','interview_score(out of 10)']],df['salary($)'])
reg.coef_,reg.intercept_
```

```python
reg.predict([[2,9,6]])
```

```python
reg.predict([[12,10,10]])
```

```python
import joblib
```

```python
joblib.dump(reg,'Salary_Prediction_Model')
```
