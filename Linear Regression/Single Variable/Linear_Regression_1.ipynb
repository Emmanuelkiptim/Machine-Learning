Importing Libraries


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn import linear_model
```

Loading the dataset


```python
df=pd.read_csv("https://raw.githubusercontent.com/codebasics/py/refs/heads/master/ML/1_linear_reg/Exercise/canada_per_capita_income.csv")
df.head()
```

#### separate the columns


```python
df_year=df.drop('per capita income (US$)', axis='columns')
```

```python
## visualize
sns.regplot(data=df, x='year',y='per capita income (US$)',marker='o')
```

```python
## train and set the model
reg= linear_model.LinearRegression()
reg.fit(df_year,df['per capita income (US$)'])
reg.coef_,reg.intercept_

```

```python
##prediction
reg.predict([[2020]])
```
