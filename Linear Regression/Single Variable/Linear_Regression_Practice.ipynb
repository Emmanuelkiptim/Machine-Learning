### Import Libraries

```python
## for data manipulation
import pandas as pd
import numpy as np
## for data visualization
import matplotlib.pyplot as plt
import seaborn as sns
## for data processing and modelling
from sklearn import linear_model
```

## Load the dataset

```python
df=pd.read_csv('https://raw.githubusercontent.com/codebasics/py/refs/heads/master/ML/1_linear_reg/homeprices.csv')
df.head()
```

Visualize the dataset

```python
fig,ax = plt.subplots()
ax.scatter(df['area'],df['price'], color='Black',marker='x')
ax.set_title('Area vs Price')
ax.set_xlabel('Area')
ax.set_ylabel("Price")


fig.show()
```

Setup and Test the model

```python
area_df=df.drop('price',axis='columns')
price=df.price
area_df
```

```python
reg=linear_model.LinearRegression()

reg.fit(area_df,price)

reg.predict([[200]])
```

```python
reg.coef_,reg.intercept_
y= reg.coef_*200 + reg.intercept_
y
```

Conduct prediction


```python
from pickle import DEFAULT_PROTOCOL
df_to_predict=pd.read_csv('https://raw.githubusercontent.com/codebasics/py/refs/heads/master/ML/1_linear_reg/areas.csv')

reg.predict(df_to_predict)
df_to_predict['Price']=reg.predict(df_to_predict)

df_to_predict['Price']=df_to_predict['Price'].astype(int)
df_to_predict
```
