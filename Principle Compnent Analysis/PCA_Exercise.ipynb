Download heart disease dataset heart.csv in Exercise folder and do following, (credits of dataset: https://www.kaggle.com/fedesoriano/heart-failure-prediction)
1. Load heart disease dataset in pandas dataframe
2. Remove outliers using Z score. Usual guideline is to remove anything that has Z score > 3 formula or Z score < -3
3. Convert text columns to numbers using label encoding and one hot encoding
4. Apply scaling
5. Build a classification model using various methods (SVM, logistic regression, random forest) and check which model gives you the best accuracy
6. Now use PCA to reduce dimensions, retrain your model and see what impact it has on your model in terms of accuracy. Keep in mind that many times doing PCA reduces the accuracy but computation is much lighter and that's the trade off you need to consider while building models in real life

```python
df.Sex.unique() ## M,F
df.ChestPainType.unique() ## ATA,NAP,ASY,TA
df.RestingECG.unique() ## Normal,ST,LVH
df.ExerciseAngina.unique() #N,Y
df.ST_Slope .unique() ## Up,Flat,Down

```

```python
df4=df.copy()
df4
```

```python
df4.Sex.replace(
    {
        'M':0,
        'F':1
    },
    inplace=True
)
df4.ChestPainType.replace(
    {
        ## ATA,NAP,ASY,TA
        'ATA':0,
        'NAP':1,
        'ASY':2,
        'TA':3
    },
    inplace=True
)
df4.RestingECG.replace(
    {
        'Normal':0,
        'ST':1,
        'LVH':2
    },
    inplace=True
)
df4.ExerciseAngina.replace(
    {
        'N':0,
        'Y':1
    },inplace=True
)
df4.ST_Slope.replace(
    {
        ## Up,Flat,Down
        'Up':0,
        'Flat':1,
        'Down':2

    },inplace=True
)
df4
```

```python
X=df4.drop('HeartDisease', axis=1)
Y=df4.HeartDisease
```

```python
from sklearn.preprocessing import StandardScaler
StandardScaler_X=StandardScaler().fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(StandardScaler_X,Y,test_size=0.3, random_state=30)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC


model_rf=RandomForestClassifier(n_estimators=40)
model_lr=LogisticRegression()
model_svm=SVC()

```

```python
from sklearn.decomposition import PCA
x_pca=PCA(0.95)
X_PCA_variable=x_pca.fit_transform(StandardScaler_X)
x_train,x_test,y_train,y_test=train_test_split(X_PCA_variable,Y,test_size=0.3, random_state=30)

model_rf.fit(x_train,y_train)
model_rf.score(x_test,y_test)


```

```python
model_lr.fit(x_train,y_train)
model_lr.score(x_test,y_test)
```

```python
model_svm.fit(x_train,y_train)
model_svm.score(x_test,y_test)
```

Support Vector Machine is the best with a score of 0.86
